```python
# asmd 함수 정의해보기

def asmd(x, y, op):
    if op == '+':
        print(f'{x} + {y} = {x+y}')
    elif op == '-':
        print(f'{x} - {y} = {x-y}')
    elif op == '*':
        print(f'{x} * {y} = {x*y}')
    elif op == '/':
        print(f'{x} / {y} = {x+y}')
```

```python
x = int(input('첫 번째 수: '))
y = int(input('두 번째 수: '))
op = input('연산자를 입력하시오')
asmd(x, y, op)
```
